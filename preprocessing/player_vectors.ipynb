```python
import pandas as pd
import numpy as np

df = pd.read_csv("outputs/player_total_value_master_enriched_final.csv")
```

```python
player_agg = df.groupby("player").agg(
    career_bat_rpa=("bat_rpa_raw", "sum"),
    career_bowl_rpa=("bowl_rpa_raw", "sum"),
    career_bat_wpa=("bat_wpa_raw", "sum"),
    career_bowl_wpa=("bowl_wpa_raw", "sum"),

    balls_faced_rpa=("balls_faced_rpa", "sum"),
    balls_faced_wpa=("balls_faced_wpa", "sum"),
    balls_bowled=("balls_bowled", "sum"),

    matches_played=("last_match_played", "count"),
    last_match_played=("last_match_played", "max")
).reset_index()

player_agg["balls_faced"] = (
    player_agg["balls_faced_rpa"] + player_agg["balls_faced_wpa"]
)
```

```python
def infer_role(row):
    bf = row["balls_faced"]
    bb = row["balls_bowled"]

    if bf >= 300 and bf >= 3 * bb:
        return "batter"
    if bb >= 300 and bb >= 3 * bf:
        return "bowler"
    if bf >= 200 and bb >= 200:
        return "allrounder"
    return "fringe"

player_agg["role"] = player_agg.apply(infer_role, axis=1)
```

```python
for col in [
    "career_bat_rpa",
    "career_bowl_rpa",
    "career_bat_wpa",
    "career_bowl_wpa"
]:
    mu = player_agg[col].mean()
    sigma = player_agg[col].std(ddof=0)
    player_agg[col + "_z"] = (player_agg[col] - mu) / sigma
```

```python
player_agg["total_value_z"] = (
    player_agg["career_bat_rpa_z"] +
    player_agg["career_bowl_rpa_z"] +
    player_agg["career_bat_wpa_z"] +
    player_agg["career_bowl_wpa_z"]
)
```

```python
player_agg["last_match_played"] = pd.to_datetime(
    player_agg["last_match_played"]
)

today = player_agg["last_match_played"].max()
player_agg["days_since_last_match"] = (
    today - player_agg["last_match_played"]
).dt.days

player_agg["active_12m"] = player_agg["days_since_last_match"] <= 365
player_agg["active_24m"] = player_agg["days_since_last_match"] <= 730
player_agg["active_36m"] = player_agg["days_since_last_match"] <= 1095
```

```python
final_cols = [
    "player",

    "career_bat_rpa",
    "career_bowl_rpa",
    "career_bat_wpa",
    "career_bowl_wpa",

    "career_bat_rpa_z",
    "career_bowl_rpa_z",
    "career_bat_wpa_z",
    "career_bowl_wpa_z",

    "total_value_z",

    "balls_faced",
    "balls_bowled",
    "matches_played",

    "role",

    "last_match_played",
    "days_since_last_match",
    "active_12m",
    "active_24m",
    "active_36m"
]

player_vectors = player_agg[final_cols]
```

```python
player_vectors.to_csv(
    "outputs/player_vectors_base.csv",
    index=False
)
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("outputs/player_vectors_base.csv")

print("Total players:", len(df))
print("\nCurrent role distribution:")
print(df["role"].value_counts())
```

```python
def infer_role_base(row):
    bf = row.get("balls_faced", 0) or 0
    bb = row.get("balls_bowled", 0) or 0

    if bf >= 300 and bf >= 3 * bb:
        return "batter"
    if bb >= 300 and bb >= 3 * bf:
        return "bowler"
    if bf >= 200 and bb >= 200:
        return "allrounder"
    return "fringe"


df["role_base"] = df.apply(infer_role_base, axis=1)

print("\nSanity check (should match current role):")
print((df["role"] == df["role_base"]).value_counts())
```

```python
fringe = df[df["role"] == "fringe"].copy()

print("Fringe count:", len(fringe))

fringe[["balls_faced", "balls_bowled"]].describe()
```

```python
fringe["bf_bb_ratio"] = (fringe["balls_faced"] + 1) / (fringe["balls_bowled"] + 1)
fringe["bb_bf_ratio"] = (fringe["balls_bowled"] + 1) / (fringe["balls_faced"] + 1)

fringe[["bf_bb_ratio", "bb_bf_ratio"]].describe()
```

```python
def infer_role_with_promotion(row):
    bf = row.get("balls_faced", 0) or 0
    bb = row.get("balls_bowled", 0) or 0

    # Primary
    if bf >= 300 and bf >= 3 * bb:
        return "batter"
    if bb >= 300 and bb >= 3 * bf:
        return "bowler"
    if bf >= 200 and bb >= 200:
        return "allrounder"

    # Secondary promotion
    # Batting specialist emerging
    if bf >= 200 and bb <= 20:
        return "batter"

    # Bowling specialist emerging
    if bb >= 200 and bf <= 20:
        return "bowler"

    return "fringe"
```

```python
df["role"] = df.apply(infer_role_with_promotion, axis=1)
```

```python
print("After update:")
print(df["role"].value_counts())
```

```python
def assign_eligibility_tier(row, active_days=450):
    # Non-fringe players are always core
    if row["role"] != "fringe":
        return "core"

    # Fringe but recently active
    if row.get("days_since_last_match", 10**9) <= active_days:
        return "active_fringe"

    # Fringe and inactive
    return "inactive_fringe"
```

```python
df["eligibility_tier"] = df.apply(assign_eligibility_tier, axis=1)
```

```python
print("\nEligibility tier counts:")
print(df["eligibility_tier"].value_counts())

# Optional cross-check
print("\nRole × Eligibility:")
print(pd.crosstab(df["role"], df["eligibility_tier"]))
```

```python
df.to_csv("outputs/player_vectors_base.csv", index=False)
print(" player_vectors_base.csv updated with eligibility_tier")
```
