```python
import pandas as pd


VENUE_MATCH_CSV = "data_processed/venue_match_level_clean.csv"
METADATA_CSV = "team_data_processed/t20_match_metadata.csv"
OUT_CSV = "team_data_processed/t20_match_metadata_filtered.csv"


venue_df = pd.read_csv(VENUE_MATCH_CSV)
meta_df = pd.read_csv(METADATA_CSV)


venue_df["match_id"] = venue_df["match_id"].astype(str)
meta_df["match_id"] = meta_df["match_id"].astype(str)


# FILTER METADATA BY MATCH IDS

valid_match_ids = set(venue_df["match_id"].unique())

filtered_meta = meta_df[meta_df["match_id"].isin(valid_match_ids)]

filtered_meta.to_csv(OUT_CSV, index=False)


print("Saved:", OUT_CSV)
print("\nCounts:")
print("Venue match ids       :", len(valid_match_ids))
print("Metadata before filter:", len(meta_df))
print("Metadata after filter :", len(filtered_meta))
```
